```python
corpus="""hwllo,welocome to nlp module. 
this is natural lanfuage processing module.
we try to implmenet nlp using nltk. """
```

```python
##tokeinsation
#sentence--> paragraph
import nltk
nltk.data.path.append('C:/Users/DELL/nltk_data')  # Optional: set a custom path
nltk.download('punkt', download_dir='C:/Users/DELL/nltk_data')
```

```python
from nltk.tokenize import sent_tokenize

document  = sent_tokenize(corpus)

print(document)
```

```python
# paragraph/sentence to wrod
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

```python
#streaming
# it is very important in nlp 
# it is a process of converting word to its word stem , 
# root leema and prefix ans suffix , increase efficency in creating vectors
words=['eat','eaten','eating','writes','writing','program','programming','history','final','finalise']


```

```python
##porter stemmer
from nltk.stem import PorterStemmer
porterstem=PorterStemmer()
for word in words:
    print(word+"---->"+ porterstem.stem(word))
```
