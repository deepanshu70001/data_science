```python
#streaming
# it is very important in nlp 
# it is a process of converting word to its word stem , 
# root leema and prefix ans suffix , increase efficency in creating vectors
#it is used in classification of word
#ex eat os root of eaten eating 
words=['eat','eaten','eating','writes','writing','program','programming','history','final','finalise']

```

```python
##porter stemmer
from nltk.stem import PorterStemmer
porterstem=PorterStemmer()
for word in words:
    print(word+"---->"+ porterstem.stem(word))
```

```python
porterstem.stem("congratulation")
#meaning of word is change --> disadvantage
```

```python
#regexpstemmer for regular expression
#we can give regular gramer stemmer algorithm
#it remove prefix and suffix acc to regular expression given

from nltk.stem import RegexpStemmer

reg_stemer=RegexpStemmer('ing$|s$|e$|able$',min=4)
```
